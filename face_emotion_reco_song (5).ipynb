```python
import numpy as np
import matplotlib.pyplot as mp
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator,load_img
from tensorflow.keras import layers,models
```

```python
td=r"C:\Users\91945\OneDrive\Desktop\emotions_song_recom\train"
vd=r"C:\Users\91945\OneDrive\Desktop\emotions_song_recom\validation"
```

```python
import os
for i in os.listdir(td):
    print(i)
```

```python
emotion=[]
for i in os.listdir(vd):
    d=[i]
    emotion.append(d)
emotion
```

```python
train_data_gen = ImageDataGenerator() # imagedatagenerator is take iamge and do augmentation on image before feeding into training
tr_dt=train_data_gen.flow_from_directory(
    td,                               # it is the directory where the image located
    shuffle=False,                                        
    target_size=(48,48),               # we are resizing image 48*48 pixel
    color_mode="grayscale",            #  image we want is black and white
    class_mode="categorical",          #    means it is like one hot encoding
    batch_size=128                     #  it give 128 image per step
)
```

```python
val_data_gen = ImageDataGenerator()
vd_dt=train_data_gen.flow_from_directory(
    vd,
    shuffle=False,
    target_size=(48,48),
    color_mode="grayscale",
    class_mode="categorical",
    batch_size=128
)
```

```python
m=models.Sequential()
m.add(layers.Conv2D(64,(3,3),activation="relu",input_shape=(48,48,1)))
m.add(layers.BatchNormalization()) # we are applying normilization
m.add(layers.MaxPooling2D(pool_size=(2,2)))
m.add(layers.Dropout(0.25))

m.add(layers.Conv2D(128,(5,5),activation="relu"))
m.add(layers.BatchNormalization()) # we are applying normilization
m.add(layers.MaxPooling2D(pool_size=(1,1)))
m.add(layers.Dropout(0.25))

m.add(layers.Conv2D(256,(3,3),activation="relu"))
m.add(layers.BatchNormalization()) # we are applying normilization
m.add(layers.MaxPooling2D(pool_size=(1,1)))
m.add(layers.Dropout(0.25))

m.add(layers.Conv2D(512,(3,3),activation="relu"))
m.add(layers.BatchNormalization()) # we are applying normilization
m.add(layers.MaxPooling2D(pool_size=(1,1)))
m.add(layers.Dropout(0.25))

m.add(layers.Flatten())

m.add(layers.Dense(128))
m.add(layers.BatchNormalization()) 
m.add(layers.Activation("relu"))
m.add(layers.Dropout(0.25))
m.add(layers.Dense(7,activation="softmax"))
m.summary()
```

```python
m.compile(optimizer="adam",loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
m.fit(tr_dt,validation_data=vd_dt,epochs=12,batch_size=16,verbose=1)
```
